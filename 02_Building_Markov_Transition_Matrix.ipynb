```python
import numpy as np
import pandas as pd

def build_models(sequence):
    nucleotides = ['A','C', 'G', 'T']
    
    zero_order_counts = {nuc: 0 for nuc in nucleotides}
    first_order_counts = {nuc1: {nuc2: 0 for nuc2 in nucleotides} for nuc1 in nucleotides}
    
    for i in range(len(sequence)):
        current_nuc = sequence[i]
        if current_nuc in nucleotides:
            zero_order_counts[current_nuc] += 1
        
        if i < len(sequence) - 1:
            next_nuc = sequence[i+1]
            if current_nuc in nucleotides and next_nuc in nucleotides:
                first_order_counts[current_nuc][next_nuc] += 1
    
    first_order_matrix = pd.DataFrame.from_dict(
        {nuc1: {nuc2: first_order_counts[nuc1][nuc2] / sum(first_order_counts[nuc1].values()) 
                for nuc2 in nucleotides} 
         for nuc1 in nucleotides},
        orient='index'
    )
    
    return first_order_matrix


#I am taking the sequence given in hidden-markov model paper of Nature Primer as the input sequence in this program.
sequence = "CTTCATGTGAAAGCAGACGTAAGTCA"

first_order = build_models(sequence)

print("1st-order Markov Model (Transition Matrix):")
print(first_order)
```
