```python
def convert_fasta_to_single_line(input_file, output_file):
    with open(input_file, 'r') as infile, open(output_file, 'w') as outfile:
        header = ""
        sequence = ""
        
        for line in infile:
            line = line.strip()  
            if line.startswith('>'):  
                if header:  
                    outfile.write(f"{header}\n{sequence}\n")
                header = line  
                sequence = ""  
            else:  
                sequence += line 
        
        if header:
            outfile.write(f"{header}\n{sequence}\n")

#Main program
input_file = "multiline_input.fasta" 
output_file = "singleline_output.fasta" 

convert_fasta_to_single_line(input_file, output_file)
print(f"Converted {input_file} to single-line format in {output_file}.")
```
